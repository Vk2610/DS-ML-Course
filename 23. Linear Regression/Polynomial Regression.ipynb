## 📖 Introduction
In Machine Learning, not all relationships between variables are linear.

Some data follows:
- Curved trends
- Parabolic relationships
- Exponential-like patterns
- Non-linear patterns

In such cases, Linear Regression performs poorly.

This is where Polynomial Regression is used.

Polynomial Regression is a supervised learning algorithm used to model non-linear relationships between independent and dependent variables by fitting a polynomial equation.

## ❓ What is Polynomial Regression?

Instead of fitting a straight line:

y = b0 + b1x

It fits:

y = b0 + b1x + b2x² + b3x³ + ... + bnxⁿ

Where:
- b₀ = intercept
- b₁..bₙ = coefficients
- x = input
- n = degree

## ⚠️ Why Linear Regression Fails Sometimes

Linear Regression assumes:

y = mx + c

But real-world data can have:
- U-shaped patterns
- Growth curves
- Curved relationships

Examples:
- Salary vs Experience
- Temperature vs Energy Consumption
- Population Growth
- Stock Trends

## ⚙️ How Polynomial Regression Works

Suppose:

X = [2,3,4]

For degree 2:

[x, x²]

Transforms into:

[2,4], [3,9], [4,16]

Then Linear Regression is applied on transformed features.

## 🔢 Types of Polynomial Degrees

Degree 1:
y = b0 + b1x

Degree 2:
y = b0 + b1x + b2x²

Degree 3:
y = b0 + b1x + b2x² + b3x³

Higher degrees:
Flexible but may overfit.

## 📌 Assumptions
- Relationship is polynomial
- Observations independent
- Errors normally distributed
- No severe multicollinearity

## ✅ Advantages
- Captures non-linear patterns
- Better fit than linear regression
- Simple implementation
- Flexible

## ❌ Disadvantages
- Can overfit
- Degree selection difficult
- Sensitive to outliers

## ⚠️ Overfitting in Polynomial Regression

Degree 1 → Underfitting  
Degree 3 → Good Fit  
Degree 15 → Overfitting

## 🛠 How to Prevent Overfitting
- Cross-validation
- Ridge/Lasso
- Optimal degree selection
- More data
- Feature scaling

## 🌍 Applications
- Sales Forecasting
- Population Growth
- Weather Modeling
- Stock Trends
- Demand Prediction

## 📈 Linear vs Polynomial Regression

| Feature | Linear | Polynomial |
|---------|--------|------------|
| Shape | Straight Line | Curve |
| Non-linearity | No | Yes |
| Complexity | Low | Higher |
| Overfitting Risk | Low | High |

## 🧾 Summary
- Extends Linear Regression
- Models curved relationships
- Uses polynomial features
- Can overfit if degree too high


```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression

np.random.seed(0)
X=np.sort(np.random.rand(30)).reshape(-1,1)
y=np.sin(2*np.pi*X).ravel()+np.random.randn(30)*0.1
X_plot=np.linspace(0,1,200).reshape(-1,1)
```

## 📊 Graphical Intuition: Different Polynomial Degrees

```python
# Degree comparison graph
degrees=[1,3,15]

plt.figure()
plt.scatter(X,y)

for d in degrees:
    model=make_pipeline(
        PolynomialFeatures(d),
        LinearRegression()
    )
    model.fit(X,y)
    plt.plot(
        X_plot,
        model.predict(X_plot),
        label=f'Degree {d}'
    )

plt.legend()
plt.title("Underfit vs Good Fit vs Overfit")
plt.show()
```

## 💻 Implementation in Python

```python
# Standard polynomial regression plot

X2=np.array([1,2,3,4,5]).reshape(-1,1)
y2=np.array([1,4,9,16,25])

poly=PolynomialFeatures(degree=2)
X_poly=poly.fit_transform(X2)

model=LinearRegression()
model.fit(X_poly,y2)

X_line=np.linspace(1,5,100).reshape(-1,1)

plt.figure()
plt.scatter(X2,y2)
plt.plot(
X_line,
model.predict(poly.transform(X_line))
)

plt.title("Polynomial Regression")
plt.show()
```
