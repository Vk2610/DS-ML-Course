# 📉 Gradient Descent Algorithm

## 📌 Introduction
**Gradient Descent** is an optimization algorithm used to minimize a function 
by iteratively moving in the direction of the steepest descent (negative gradient).

It is widely used in machine learning to minimize cost functions.


## ⚙️ How Gradient Descent Works

1. Initialize parameters randomly  
2. Compute gradient of cost function  
3. Update parameters using rule:

θ = θ - α * (dJ/dθ)

4. Repeat until convergence  

### Terms:
- θ → parameters  
- α → learning rate  
- J → cost function  


```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Cost function and gradient

def cost_function(x):
    return x**2  # simple convex function

def gradient(x):
    return 2*x
```

```python
# Gradient Descent Simulation

x = 8  # initial value
learning_rate = 0.1
iterations = 25

x_vals = []
cost_vals = []

for i in range(iterations):
    x_vals.append(x)
    cost_vals.append(cost_function(x))
    x = x - learning_rate * gradient(x)

print("Final value:", x)
```

```python
# Plot Cost vs Iterations

plt.plot(cost_vals)
plt.title("Cost Reduction (Convergence)")
plt.xlabel("Iterations")
plt.ylabel("Cost")
plt.show()
```

```python
# Plot Gradient Descent Path

x_range = np.linspace(-10, 10, 100)
y_range = x_range**2

plt.plot(x_range, y_range)
plt.scatter(x_vals, [cost_function(x) for x in x_vals])

plt.title("Gradient Descent Steps")
plt.xlabel("x")
plt.ylabel("Cost")
plt.show()
```
