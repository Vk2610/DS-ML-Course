# 📉 Convergence Algorithm (Gradient Descent)

## 📌 Introduction
A convergence algorithm iteratively finds an optimal solution.
Gradient Descent minimizes a cost function step-by-step until convergence.

## ⚙️ Details

Update rule:
θ = θ - α * (dJ/dθ)

Repeat until cost stabilizes.


```
import numpy as np
import matplotlib.pyplot as plt
```

```
def cost_function(x):
    return x**2

def gradient(x):
    return 2*x

x = 10
lr = 0.1
iterations = 20

x_vals = []
cost_vals = []

for i in range(iterations):
    x_vals.append(x)
    cost_vals.append(cost_function(x))
    x = x - lr * gradient(x)

plt.plot(cost_vals)
plt.xlabel("Iterations")
plt.ylabel("Cost")
plt.title("Convergence Graph")
plt.show()
```

```
x_range = np.linspace(-10, 10, 100)
y_range = x_range**2

plt.plot(x_range, y_range)
plt.scatter(x_vals, [cost_function(x) for x in x_vals])
plt.xlabel("x")
plt.ylabel("Cost")
plt.title("Gradient Descent Path")
plt.show()
```
