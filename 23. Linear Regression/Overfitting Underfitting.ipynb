
# 📉 Overfitting and Underfitting in Machine Learning

## 📖 Introduction
In Machine Learning, the goal is to build models that generalize well to unseen data. However, models often fail in two major ways:

- Overfitting → Model learns too much (including noise)  
- Underfitting → Model learns too little  

Understanding these concepts is crucial for building accurate and robust models.

---

## ⚠️ What is Overfitting?
Overfitting occurs when a model learns the training data too well, including noise and outliers.

### 🔍 Characteristics:
- Very high accuracy on training data  
- Poor performance on unseen/test data  
- Model is too complex  

### 🧠 Intuition:
The model memorizes instead of learning patterns.

---

## ❌ What is Underfitting?
Underfitting occurs when a model is too simple to capture the underlying patterns in the data.

### 🔍 Characteristics:
- Poor performance on both training and test data  
- Model fails to learn relationships  
- High bias  

### 🧠 Intuition:
The model is too naive to understand the data.

---

## ⚖️ Bias vs Variance Tradeoff

| Concept | Description |
|--------|------------|
| Bias | Error due to overly simple assumptions |
| Variance | Error due to sensitivity to small fluctuations |

| Case | Bias | Variance |
|------|------|----------|
| Underfitting | High | Low |
| Overfitting | Low | High |
| Ideal Model | Balanced | Balanced |

---

## 📊 Visual Intuition

- 🔹 Underfitting (Too Simple Model)
- 🔹 Good Fit (Balanced Model)
- 🔹 Overfitting (Too Complex Model)

---

## ⚠️ Causes of Overfitting and Underfitting

### 🔴 Causes of Overfitting:
- Too complex model (e.g., high-degree polynomial)  
- Too many features  
- Small training dataset  
- Noise in data  
- No regularization  

### 🔵 Causes of Underfitting:
- Model too simple  
- Insufficient training time  
- Poor feature selection  
- Low model capacity  

---

## 🔍 How to Detect Them

| Metric | Underfitting | Overfitting |
|--------|-------------|-------------|
| Training Error | High | Very Low |
| Test Error | High | High |
| Gap (Train vs Test) | Small | Large |

---

## 🛠️ Techniques to Prevent Overfitting

### ✅ 1. Cross-Validation
Use techniques like K-Fold Cross Validation  

### ✅ 2. Regularization
- L1 (Lasso)  
- L2 (Ridge)  

### ✅ 3. Reduce Model Complexity
- Lower polynomial degree  
- Prune decision trees  

### ✅ 4. More Training Data
Helps model generalize better  

### ✅ 5. Dropout (Neural Networks)
Randomly deactivate neurons  

### ✅ 6. Early Stopping
Stop training before overfitting occurs  

---

## 🔧 Techniques to Fix Underfitting

- Increase model complexity  
- Add more relevant features  
- Train longer  
- Reduce regularization  
- Use better algorithms  

---

## 🧾 Summary

| Aspect | Underfitting | Overfitting |
|--------|-------------|-------------|
| Model Complexity | Low | High |
| Training Accuracy | Low | High |
| Test Accuracy | Low | Low |
| Bias | High | Low |
| Variance | Low | High |


```python

from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.pipeline import make_pipeline
import numpy as np
import matplotlib.pyplot as plt

# Sample Data
X = np.array([1,2,3,4,5]).reshape(-1,1)
y = np.array([1,4,9,16,25])

# Underfitting Model (Linear)
model1 = LinearRegression()
model1.fit(X, y)

# Overfitting Model (High-degree polynomial)
model2 = make_pipeline(PolynomialFeatures(10), LinearRegression())
model2.fit(X, y)

# Plot
x_plot = np.linspace(1,5,100).reshape(-1,1)

plt.figure()
plt.scatter(X, y)
plt.plot(x_plot, model1.predict(x_plot), label="Underfit")
plt.plot(x_plot, model2.predict(x_plot), label="Overfit")
plt.legend()
plt.title("Underfitting vs Overfitting")
plt.show()
```
