# 📈 Linear Regression (with Visual Explanation)

## 📌 What is Linear Regression?

**Linear Regression** is a supervised machine learning algorithm used to model the relationship between a **dependent variable (target)** and one or more **independent variables (features)**.

It assumes a **linear relationship** between input and output.

---

## 🧠 Types of Linear Regression

### 1. Simple Linear Regression

**Equation:**
( y = mx + c )

* One input variable
* Straight line relationship

---

### 2. Multiple Linear Regression

**Equation:**
( y = b_0 + b_1x_1 + b_2x_2 + \dots + b_nx_n )

* Multiple input variables
* Forms a **plane (or hyperplane)**

---

## ⚙️ How Linear Regression Works (Step-by-Step)

### 🔄 Training Process

1. Initialize random line
2. Predict values
3. Calculate error
4. Adjust slope & intercept
5. Repeat until error is minimized

---

---

## 📉 Cost Function (MSE)

[
MSE = \frac{1}{n} \sum (y_i - \hat{y}_i)^2
]

### 🔻 Cost Reduction Graph

```
Error
│\
│ \
│  \
│   \
│    \_____
│___________ Iterations
```

👉 Error decreases as model improves

## 🧪 Python Example with Graph

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Data
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([2, 4, 6, 8, 10])

# Model
model = LinearRegression()
model.fit(X, y)

# Prediction
y_pred = model.predict(X)

# Plot
plt.scatter(X, y)
plt.plot(X, y_pred)
plt.title("Linear Regression Fit")
plt.xlabel("X")
plt.ylabel("y")
plt.show()
```

## 📌 Overview
This notebook demonstrates **Linear Regression** with:
- Data visualization
- Model training
- Best fit line
- Residuals (error visualization)


````python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
````

````python
# Sample data
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([2, 4, 6, 8, 10])

plt.scatter(X, y)
plt.title("Original Data")
plt.xlabel("X")
plt.ylabel("y")
plt.show()
````

````python
# Train model
model = LinearRegression()
model.fit(X, y)

# Predictions
y_pred = model.predict(X)

print("Slope:", model.coef_[0])
print("Intercept:", model.intercept_)
````

````python
# Plot regression line
plt.scatter(X, y, label="Actual Data")
plt.plot(X, y_pred, label="Best Fit Line")
plt.title("Linear Regression Fit")
plt.xlabel("X")
plt.ylabel("y")
plt.legend()
plt.show()
````

````python
# Residuals (errors)
residuals = y - y_pred

plt.scatter(X, residuals)
plt.axhline(y=0)
plt.title("Residual Plot")
plt.xlabel("X")
plt.ylabel("Error")
plt.show()
````
