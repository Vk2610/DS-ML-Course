# Tuples

A tuple is a collection of ordered, immutable elements. Tuples are defined using parentheses `()` and can contain elements of different data types. Once a tuple is created, its contents cannot be modified.

````python
## creating a tuple 
tuple = ()
print(type(tuple))
my_tuple = (1, 2, 3, 'hello', 4.5)
print(my_tuple)
````

````python
lst = list()
print(type(lst))
````

````python
numbers = tuple([1, 2, 3, 4, 5])
print(numbers)
````

## Tuples and list can be converted to each other

```python
# creating a tuple
tuple = ()
print(type(tuple))
my_tuple = (1, 2, 3, 'hello', 4.5)
print(my_tuple)
# creating a list
lst = list()
print(type(lst))
# converting list to tuple
numbers = tuple([1, 2, 3, 4, 5])
print(numbers)
```


````python
## creating a tuple
tuple = ()
print(type(tuple))
my_tuple = (1, 2, 3, 'hello', 4.5)
print(my_tuple)

# Accessing tuple elements
print(my_tuple[0])  # Output: 1
print(my_tuple[3])  # Output: 'hello'
````

````python
## Tuples operations
# Concatenation
tuple1 = (1, 2, 3)
tuple2 = ('a', 'b', 'c')
concatenated_tuple = tuple1 + tuple2
print(concatenated_tuple)  # Output: (1, 2, 3, 'a', 'b', 'c')
````

````python
# Repetition
repeated_tuple = tuple1 * 2
print(repeated_tuple)  # Output: (1, 2, 3, 1, 2, 3)
````

````python
## Immutable nature of tuples

list = [1, 2, 3]
tuple = (1, 2, 3)
# Modifying the list
list[0] = 10
print(list)  # Output: [10, 2, 3]
# Attempting to modify the tuple (this will raise an error)
try:
    tuple[0] = 10
except TypeError as e:
    print(e)  # Output: 'tuple' object does not support item assignment
````

````python
## Tuple methods 
my_tuple = (1,1,1,1,1,1, 2, 3, 'hello', 4.5)
# Count method
print(my_tuple.count(1))  # Output: 6
print(my_tuple.count('hello'))  # Output: 1
````

````python
# Index method
print(my_tuple.index(2))  # Output: 6
````

````python
# Packing and unpacking of tuples

# Packing
packed_tuple = 1, 'hello', 3.14
print(packed_tuple)  # Output: (1, 'hello', 3.14)
````

````python
# Unpacking tuples
a, b, c = packed_tuple
print(a)  # Output: 1
print(b)  # Output: 'hello'
print(c)  # Output: 3.14
````

````python
numbers = (1, 2, 3, 4, 5)
first, *middle, last = numbers
print(first)   # Output: 1
print(middle)  # Output: [2, 3, 4]
print(last)    # Output: 5
````

````python
## Nested tuples
nested_tuple = (1, 2, (3, 4), 5)
print(nested_tuple)  # Output: (1, 2, (3, 4), 5)
print(nested_tuple[2])  # Output: (3, 4)
````

````python
# Nested lists
nested_list = [1, 2, [3, 4], 5]
print(nested_list)  # Output: [1, 2, [3, 4], 5]
print(nested_list[2])  # Output: [3, 4]
````

````python
print(nested_list[2][1])
````

````python
nested_tuple =((1,2,3),('a','b','c'), (True, False))

# Accessing nested tuple elements
print(nested_tuple[0])
print(nested_tuple[1][2])
````

````python
# iterating over nested tuple
for inner_tuple in nested_tuple:
    for element in inner_tuple:
        print(element)
        
````

## Conclusion 
Tuples are a powerful data structure in Python that allow you to store and manipulate ordered collections of data. They are immutable, which means that once a tuple is created, its contents cannot be changed. This makes tuples useful for storing data that should not be modified, such as coordinates or fixed values. Tuples can also be nested, allowing you to create complex data structures. Additionally, tuples can be converted to lists and vice versa, providing flexibility in how you work with your data.
