### Example 1. Manage a TO-DO list
 Create a To Do list to keep track of tasks

```python
to_do_list = ['Buy groceries', 'Clean the house', 'Finish homework']

# Add a new task to the list
to_do_list.append('Pay bills')

# Remove a completed task
to_do_list.remove('Clean the house')

# check if a task is in the list
if 'Finish homework' in to_do_list:
    print("Don't forget to finish your homework!")
    
# Display the number of remaining tasks in the To Do list
print(f"You have {len(to_do_list)} tasks remaining.")

# Print the updated To Do list
print(to_do_list)
```

### Example 2. Organizing student grades
 Create a list of student names and their corresponding grades

```python
# organinzing student grades
students = ['Alice', 'Bob', 'Charlie']
grades = [85, 92, 78]

# Create a list of tuples pairing each student with their grade
student_grades = list(zip(students, grades))
# Print the list of student grades
print(student_grades)

# Calculate the average grade
average_grade = sum(grades) / len(grades)

print(f"The average grade is: {average_grade}")

# Find the student with the highest grade
highest_grade = max(grades)
highest_grade_index = grades.index(highest_grade)
top_student = students[highest_grade_index]
print(f"The top student is {top_student} with a grade of {highest_grade}.") 
```

### Example 3. Inventory management
 Create a list of products in a store and their quantities

```python
#  managing inventory
products = ['Apples', 'Bananas', 'Oranges']
quantities = [50, 100, 75]

# add new product to the inventory
products.append('Grapes')
quantities.append(60)


#  checking if an item is in stock
item_to_check = 'Bananas'
if item_to_check in products:
    index = products.index(item_to_check)
    print(f"{item_to_check} is in stock with quantity: {quantities[index]}")
else:
    print(f"{item_to_check} is not in stock.")  

# Update the quantity of a product
product_to_update = 'Oranges'
new_quantity = 80
if product_to_update in products:
    index = products.index(product_to_update)
    quantities[index] = new_quantity
    print(f"Updated {product_to_update} quantity to {new_quantity}.")
    
# Display the current inventory
inventory = list(zip(products, quantities))
print("Current Inventory:")
for product, quantity in inventory:
    print(f"{product}: {quantity}")
    
```

### Example 4. Collecting user feedback
  Create a list to store user feedback for a product or service

```python
# Collecting user feedback
feedback = ['Greate service!', 'Could be better.', 'Loved the product!', 'Not satisfied with the quality.', 'Will buy again!']

# Add new feedback
feedback.append('Customer support was helpful.')

# Analyze feedback
positive_feedback = [f for f in feedback if 'great' in f.lower() or 'loved' in f.lower() or 'helpful' in f.lower()]
negative_feedback = [f for f in feedback if 'not satisfied' in f.lower() or 'could be better' in f.lower()] 
print(f"Positive feedback: {len(positive_feedback)}")
print(f"Negative feedback: {len(negative_feedback)}") 
```
