# Sets in Python

A set is an unordered collection of unique elements. It is defined using curly braces `{}` or the `set()` constructor. Sets are mutable, meaning you can add or remove elements from them.

```python
## create a set 
my_set = {1, 2, 3, 4, 5}
print(my_set)
print(type(my_set))
```

```python
#  empty set
empty_set = set()
print(empty_set)
print(type(empty_set))
```

```python
empty_set = set([1, 2, 3, 4, 5, 6])
print(empty_set)
```

```python
# Sets do not allow duplicate values
my_set = {1, 2, 3, 4, 5, 5, 5}
print(my_set)
```

```python
#  Basic set operations
set_a = {1, 2, 3, 4, 5}
set_b = {4, 5, 6, 7, 8}
# Union
union_set = set_a.union(set_b)
print("Union:", union_set)
```

```python
# Adding an element to a set
my_set = {1, 2, 3}
my_set.add(4)
print(my_set)
```

```python
# removing an element from a set  
my_set.remove(2)
print(my_set)
```

```python
# discard an element from a set (does not raise an error if the element is not present)
my_set.discard(10)  # This will not raise an error
```

```python
# pop an element from a set (removes and returns an arbitrary element)
popped_element = my_set.pop()
print("Popped element:", popped_element)
print("Set after popping an element:", my_set)
```

```python
#  Set membership test
print(3 in my_set)  # True
print(10 in my_set)  # False
```

```python
#  mathematical set operations
set_a = {1, 2, 3, 4, 5}
set_b = {4, 5, 6, 7, 8}

# union 
union_set = set_a.union(set_b)
print("Union:", union_set)
# Intersection
intersection_set = set_a.intersection(set_b)
print("Intersection:", intersection_set)

#  intersection update
set_a.intersection_update(set_b)
print("Set A after intersection update:", set_a)
```

```python
# difference
difference_set = set_a.difference(set_b)
print("Difference (A - B):", difference_set)
```

```python
# Symmetric difference
sym_diff_set = set_a.symmetric_difference(set_b)
print("Symmetric Difference:", sym_diff_set)
```

```python
# Sets methods 
set1 = {1, 2, 3}
set2 = {3, 4, 5}
# issubset
print(set1.issubset(set2))  # False

# issuperset
print(set1.issuperset(set2))  # False
```

```python
lt = [1, 2, 3, 4, 5,5,6,4]
st = set(lt)
print(st)
```

```python
# counting unique elements in a list using set
my_list = [1, 2, 3, 4, 5, 5, 6, 4]
unique_elements = set(my_list)  
print(unique_elements)
```

```python
# in text 
text = 'in this tutorial we will learn about sets in python'
words = text.split()
unique_words = set(words)
print(unique_words)
print(len(unique_words))
```
