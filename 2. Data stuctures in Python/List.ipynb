# Lists in Python

Lists are a fundamental data structure in Python that allow you to store and manage collections of items. They are ordered, mutable, and can contain elements of different types. In this section, we will explore the basics of lists, how to create them, and some common operations you can perform on them.
## Creating Lists
You can create a list by enclosing a comma-separated sequence of items in square brackets `[]`.
```python# Creating a list of integers
numbers = [1, 2, 3, 4, 5]
# Creating a list of strings
fruits = ["apple", "banana", "cherry"]
# Creating a mixed list
mixed = [1, "hello", 3.14, True]
```

## Accessing List Elements
You can access individual elements of a list using their index. The index starts at 0 for the first element.
```python# Accessing elements
print(numbers[0])  # Output: 1
print(fruits[1])  # Output: banana
print(mixed[2])   # Output: 3.14
```

## Modifying Lists
Lists are mutable, which means you can change their contents after they have been created.
```python# Modifying an element
numbers[0] = 10
print(numbers)  # Output: [10, 2, 3, 4, 5]
# Adding an element
fruits.append("orange")
print(fruits)  # Output: ['apple', 'banana', 'cherry', 'orange']
# Removing an element
mixed.remove("hello")
print(mixed)  # Output: [1, 3.14, True]
```
## List Operations
Python provides various built-in functions and methods to perform operations on lists.
```python# Length of a list 
print(len(numbers))  # Output: 5
# Concatenating lists
combined = numbers + fruits
print(combined)  # Output: [10, 2, 3, 4, 5, 'apple', 'banana', 'cherry', 'orange']
# Repeating lists
repeated = fruits * 2
print(repeated)  # Output: ['apple', 'banana', 'cherry', 'orange', 'apple', 'banana', 'cherry', 'orange']
```
## List Slicing
You can extract a portion of a list using slicing.
```python# Slicing a list
print(numbers[1:4])  # Output: [2, 3, 4]
print(fruits[:2])    # Output: ['apple', 'banana']
print(mixed[1:])    # Output: [3.14, True]
```

## Conclusion
Lists are a versatile and essential data structure in Python. They allow you to store and manipulate collections of items efficiently. Understanding how to create, access, modify, and perform operations on lists is crucial for any Python programmer.

#### List comprehension

Basic syntax of list comprehension is:
```pythonnew_list = [expression for item in iterable if condition]
```

with condition is optional. It allows you to create a new list by applying an expression to each item in an iterable, optionally filtering items using a condition.

Nested list comprehension is a powerful tool that allows you to create complex lists in a concise and readable way. It involves using a list comprehension inside another list comprehension. This can be useful for creating multi-dimensional lists or for applying multiple transformations to the items in a list.

For example, if you want to create a 2D list (a list of lists) that contains the squares of numbers from 0 to 4, you can use nested list comprehension like this:
```pythonsquared_2d = [[x**2 for x in range(5)] for _ in range(3)]  
print(squared_2d)
```
This will output:
```
[[0, 1, 4, 9, 16], 
 [0, 1, 4, 9, 16], 
 [0, 1, 4, 9, 16]]
```

````python
## nested list comprehension
matrix = [[1,2,3],[4,5,6],[7,8,9]]
flattened = [num for row in matrix for num in row]
print(flattened)
````

````python
list1 = [1,2,3,4]
list2 = ['a','b','c','d']
paired = [(x,y) for x in list1 for y in list2]
print(paired)
print(len(paired))
````

````python
## List comprehension with function calls

numbers = [1,2,3,4,5]
def square(x):
    return x**2
squared_numbers_func = [square(x) for x in numbers]
print(squared_numbers_func) 
````
