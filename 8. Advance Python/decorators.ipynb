# Decorators in Python

Decorators are a powerful and flexible feature in python that allows you to modify the behavior of a functions or class method. They are commonly used to add functionality to the functions or methods without modifying their actual code.

```python
# function copy 

def welcome():
  return "Welcome to Python"

print(welcome())
wel=welcome
print(wel())

del welcome
print(wel())
```

```python
# closures
#  function inside function

def main_welcome():
  msg="Welcome to Python"
  def sub_welcome_method():
    print("Welcome to advanced python course")
    print(msg)
    print("Learn the concepts properly")
    
  return sub_welcome_method

welcome=main_welcome()
welcome()
```

```python
# creating a decorator 

def main_welcome(func):   # must accept function as argument
    def sub_welcome_method():
        print("Welcome to advanced python course")
        func()   # calling original function
        print("Learn the concepts properly")
    return sub_welcome_method


@main_welcome 
def welcome():
    print("Welcome to Python")

welcome()
```
