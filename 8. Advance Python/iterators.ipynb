# Iterators 

- Iterators are advanced Python concepts that allow for efficient looping and memory management. Iterators provide way to access elements of a collection sequentially without exposing the underlying structure.

```python
my_list = [1,2,3,4,5]
for i in my_list:
  print(i)
```

```python
# Iterator 

iterator = iter(my_list)
print(iterator)
```

```python
#  Iterate thorugh all the items in the list
print(next(iterator))
print(next(iterator))
print(next(iterator))
print(next(iterator))
print(next(iterator))
```

```python
iterator=iter(my_list)

try:
  print(next(iterator))
except StopIteration:
  print("No more elements")
```
