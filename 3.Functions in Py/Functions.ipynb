# Functions
Functions are reusable blocks of code that perform a specific task. They allow you to break down complex problems into smaller, more manageable pieces. In programming, functions can take inputs (called parameters) and return outputs (called return values).
## Defining a Function
To define a function, you typically use a specific syntax depending on the programming language. Here’s an example in Python:

```pythondef greet(name):
    return f"Hello, {name}!"
```In this example, we define a function called `greet` that takes one parameter, `name`, and returns a greeting message.
## Calling a FunctionTo call a function, you simply use its name followed by parentheses, and pass any required arguments. For example:

```pythonmessage = greet("Alice")
print(message)  # Output: Hello, Alice!
```In this case, we call the `greet` function with the argument "Alice" and store the returned message in the variable `message`, which is then printed to the console.
## Function Parameters and ArgumentsParameters are the variables that are defined in the function definition, while arguments are the actual values that you pass to the function when you call it. For example, in the `greet` function,`name` is a parameter, and "Alice" is an argument.
## Return ValuesA function can return a value using the `return` statement. This allows you to capture the output of the function and use it elsewhere in your code. For example, in the `greet` function, we return a string that contains the greeting message.
## Benefits of Using Functions
- **Reusability**: Functions allow you to reuse code, which can save time and reduce errors.
- **Modularity**: Functions help break down complex problems into smaller, more manageable pieces.
- **Readability**: Functions can make your code more organized and easier to read.
- **Maintainability**: Functions can make it easier to maintain and update your code, as you can modify a function in one place rather than changing code in multiple locations.
## ConclusionFunctions are a fundamental concept in programming that enable you to write more efficient, organized, and maintainable code. By defining and using functions, you can improve the structure of your code and make it easier to understand and debug.

````python
def even_or_odd(num):
    '''This function checks if a number is even or odd.'''
    if num%2==0:
        return f"{num} is even"
    else:    
        return f"{num} is odd"  
print(even_or_odd(23))
````

````python
# Functions with multiple parameters
def add_numbers(num1, num2):
    '''This function adds two numbers and returns the result.'''
    return num1 + num2
print(add_numbers(5, 7))
````

````python
# default parameters
# If a parameter has a default value, it becomes optional when calling the function. If the caller does not provide a value for that parameter, the default value will be used.
# Functions with default parameters
def greet(name, greeting="Hello"):
    '''This function greets a person with a given greeting.'''
    return f"{greeting}, {name}!"
print(greet("Alice"))
print(greet("Bob", "Hi"))
````

````python
# Variable length arguments
# Sometimes, you may want to create a function that can accept a variable number of arguments. In Python, you can use *args for non-keyword variable length arguments and **kwargs for keyword variable length arguments.
# Function with variable length arguments

def sum_all(*args):
    '''This function sums all the numbers provided as arguments.'''
    return sum(args)
print(sum_all(1, 2, 3, 4, 5))
````

````python
# Positional arguments - *args allows you to pass a variable number of non-keyword arguments to a function. Inside the function, args is treated as a tuple containing all the arguments passed.
def print_numbers(*krish):
  for num in krish:
    print(num)

print_numbers(10, 20, 30)
````

````python
# Keyword arguments - **kwargs allows you to pass a variable number of keyword arguments to a function. Inside the function, kwargs is treated as a dictionary containing all the keyword arguments passed.
def print_info(**kwargs):
    '''This function prints the information provided as keyword arguments.'''
    for key, value in kwargs.items():
        print(f"{key}: {value}")
print_info(name="Alice", age=30, city="New York")
````

````python
# mixed arguments *args and **kwargs can be used together in a function definition. When using both, *args must be placed before **kwargs in the function definition.
def mixed_arguments(*args, **kwargs):
    '''This function demonstrates the use of both *args and **kwargs.'''
    print("Positional arguments (args):")
    for arg in args:
        print(arg)
    
    print("\nKeyword arguments (kwargs):")
    for key, value in kwargs.items():
        print(f"{key}: {value}")
mixed_arguments(1, 2, 3, name="Alice", age=30)
````

````python
# Return statement - The return statement is used to exit a function and return a value to the caller. A function can have multiple return statements, but only one will be executed based on the flow of the function.

def check_number(num):
    '''This function checks if a number is positive, negative, or zero.'''
    if num > 0:
        return "Positive"
    elif num < 0:
        return "Negative"
    else:
        return "Zero"
print(check_number(10))
print(check_number(-5))
````
