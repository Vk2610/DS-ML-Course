# Lambda Function

A Lambda function is a small, anonymous function that can be defined in a single line of code. It is often used for short, simple operations that can be easily expressed in a single expression. Lambda functions are commonly used in programming languages like Python, JavaScript, and C#.

## Syntax
In Python, the syntax for a Lambda function is as follows:

lambda arguments: expression


```python
def addition(x, y):
    return x + y
print(addition(5, 3))
```

```python
add = lambda x, y: x + y
print(add(15, 32))
type(add)
```

```python
def even_or_odd(num):
    return "Even" if num % 2 == 0 else "Odd"
print(even_or_odd(10))

# Using Lambda function for even or odd
even_or_odd_lambda = lambda num: "Even" if num % 2 == 0 else "Odd"
print(even_or_odd_lambda(15))
```

```python
mult= lambda x,y,z : x*y*z
print(mult(2,3,4))
```

```python
# map() - applies a given function to each item of an iterable (like a list) and returns a list of the results.

numbers = [1, 2, 3, 4, 5]
def square(x):
    return x ** 2
squared_numbers = list(map(square, numbers))
print(squared_numbers)
```
