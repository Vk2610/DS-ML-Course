```python
# Temprature conversion function

def convert_temprature(temp,unit):
  """This function converts temperature between Celsius and Fahrenheit."""
  if unit == "C":
    return (temp * 9/5) + 32
  elif unit == "F":
    return (temp - 32) * 5/9
  else:
    return "Invalid unit. Please use 'C' for Celsius or 'F' for Fahrenheit."
print(convert_temprature(100, "C"))
print(convert_temprature(212, "F"))
```

```python
# Example 2. Password strength checker function

def check_password_strength(password):
    """This function checks the strength of a password."""
    if len(password) < 8:
        return "Weak password: Too short."
    elif not any(char.isdigit() for char in password):
        return "Weak password: No digits."
    elif not any(char.isupper() for char in password):
        return "Weak password: No uppercase letters."
    elif not any(char.islower() for char in password):
        return "Weak password: No lowercase letters."
    else:
        return "Strong password."
print(check_password_strength("password"))
print(check_password_strength("Password123"))
```

```python
# Example 3. Calculate the total cost of items in a shopping cart

def calculate_total_cost(cart):
  total_cost = 0
  for item in cart:
    total_cost += item['price'] * item['quantity']  
  return total_cost


# cart data 
cart =[
  {'name': 'Laptop', 'price': 999.99, 'quantity': 1},
  {'name': 'Headphones', 'price': 199.99, 'quantity': 2},
  {'name': 'Mouse', 'price': 49.99, 'quantity': 1}
]

print(calculate_total_cost(cart))
```

```python
# Example 4. Check if a string is a palindrome

def is_palindrome(s):
    """This function checks if a string is a palindrome."""
    cleaned_s = ''.join(s.split()).lower()  # Remove spaces and convert to lowercase
    return cleaned_s == cleaned_s[::-1]  # Check if the string is equal to its reverse
print(is_palindrome("A man a plan a canal Panama"))
print(is_palindrome("Hello World"))
```

```python
# Example 5. Calculate the factorial of a number using recursion

def factorial(n):
    """This function calculates the factorial of a number using recursion."""
    if n < 0:
        return "Factorial is not defined for negative numbers."
    elif n == 0 or n == 1:
        return 1
    else:
        return n * factorial(n - 1)
print(factorial(5))
print(factorial(0))
```
