# Log-Normal Distribution

## 📌 Introduction
A **Log-Normal Distribution** is a probability distribution of a random variable whose **logarithm is normally distributed**.

👉 In simple terms:  
If `Y = log(X)` follows a normal distribution, then `X` follows a log-normal distribution.

---

## 📊 Definition

A random variable `X` is log-normally distributed if:

ln(X) ~ N(μ, σ²)


Where:
- `μ` = mean of the logarithm of X  
- `σ²` = variance of the logarithm of X  

---

## 📐 Probability Density Function (PDF)

f(x) = (1 / (xσ√(2π))) * e^(-(ln(x) - μ)² / (2σ²)), for x > 0


---

## 📉 Graph Representation
- Skewed to the **right (positively skewed)**
- Starts from **x > 0** (never negative)
- Long tail on the right side

---

## 📌 Key Properties

### Mean:

E(X) = e^(μ + σ²/2)


### Median:

Median = e^μ


### Mode:

Mode = e^(μ - σ²)


### Variance:

Var(X) = (e^(σ²) - 1) * e^(2μ + σ²)


---

## ⚡ Characteristics
- Values are **strictly positive (X > 0)**
- Distribution is **right-skewed**
- Suitable for modeling **multiplicative processes**
- Log transformation converts it into a normal distribution

---

## 📍 Real-Life Examples
- Stock prices in finance  
- Income distribution in economics  
- Growth of bacteria/population  
- Time to complete tasks  
- Distribution of file sizes on the internet  

---

## 🔄 Relationship with Normal Distribution
- If `X` is log-normal → `ln(X)` is normal  
- If `Y` is normal → `e^Y` is log-normal  

---

## 🆚 Log-Normal vs Normal Distribution

| Feature              | Log-Normal Distribution     | Normal Distribution       |
|---------------------|-----------------------------|---------------------------|
| Range               | (0, ∞)                      | (-∞, ∞)                   |
| Shape               | Right-skewed                | Symmetrical               |
| Values              | Only positive               | Can be negative or positive |
| Transformation      | Log → Normal                | Already Normal            |

---

## ✅ Summary
- Log-Normal Distribution is used when data is **positively skewed**
- Defined by the property: **log(X) is normally distributed**
- Common in **finance, biology, and real-world growth processes**
