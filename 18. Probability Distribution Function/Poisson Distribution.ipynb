# 📊 Poisson Distribution

## 📌 Overview
The **Poisson Distribution** is a discrete probability distribution that models the number of events occurring in a fixed interval of time, space, or area, given that these events occur with a known constant mean rate and independently of the time since the last event.

---

## 🧠 Key Characteristics

- Events occur independently  
- Average rate (λ) is constant  
- Events cannot occur simultaneously  
- Suitable for rare events  

---

## 🧩 Visual Representation of Poisson Process

Time Interval ─────────────────────────────>

Events: • • • • •
↑ ↑ ↑ ↑ ↑
Random occurrences with constant average rate (λ)


👉 Events occur randomly but with a predictable average rate.

---

## 📐 Probability Mass Function (PMF)

\[
P(X = k) = \frac{e^{-\lambda} \lambda^k}{k!}
\]

Where:
- `λ` (lambda) = average number of events  
- `k` = number of occurrences  
- `e` ≈ 2.718  

---

## 📊 Mean and Variance

- **Mean:**  
  \[
  \mu = \lambda
  \]

- **Variance:**  
  \[
  \sigma^2 = \lambda
  \]

- **Standard Deviation:**  
  \[
  \sigma = \sqrt{\lambda}
  \]

---

## 🎯 Example

A call center receives **3 calls per minute** on average.  
Find probability of receiving exactly **2 calls** in a minute.

\[
P(X = 2) = \frac{e^{-3} \cdot 3^2}{2!}
\]

\[
P(X = 2) = \frac{e^{-3} \cdot 9}{2} \approx 0.224
\]

---

```python
## 📈 Poisson Distribution Graph (Embedded Image)

### 🔹 Step 1: Generate and Save Graph Image

import matplotlib.pyplot as plt
from scipy.stats import poisson

lam = 3

x = list(range(0, 15))
y = [poisson.pmf(i, lam) for i in x]

plt.figure(figsize=(8,5))
plt.bar(x, y)

plt.xlabel("Number of Events (k)")
plt.ylabel("Probability P(X = k)")
plt.title("Poisson Distribution (λ = 3)")

plt.grid(axis='y', linestyle='--', alpha=0.7)

# Save image
plt.savefig("poisson_plot.png")
plt.show()
```

```python
import matplotlib.pyplot as plt
from scipy.stats import poisson

lam_values = [1, 4, 10]

x = list(range(0, 20))

plt.figure(figsize=(10,6))

for lam in lam_values:
    y = [poisson.pmf(i, lam) for i in x]
    plt.plot(x, y, marker='o', label=f"λ={lam}")

plt.xlabel("Number of Events")
plt.ylabel("Probability")
plt.title("Poisson Distribution for Different λ Values")
plt.legend()
plt.grid()

# Save image
plt.savefig("poisson_variations.png")
plt.show()
```
