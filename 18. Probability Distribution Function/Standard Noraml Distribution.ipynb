# 📊 Standard Normal Distribution (Z-Distribution)

## 📌 Overview
The **Standard Normal Distribution** is a special case of the normal distribution where:
- Mean (μ) = 0  
- Standard Deviation (σ) = 1  

It is also known as the **Z-distribution** and is widely used in statistics for standardizing data and calculating probabilities.

---

## 🧠 Key Characteristics

- Mean = 0  
- Standard deviation = 1  
- Symmetric bell-shaped curve  
- Total area under curve = 1  
- Used to compute **Z-scores**

## 📐 Probability Density Function (PDF)

\[
f(z) = \frac{1}{\sqrt{2\pi}} e^{-\frac{z^2}{2}}
\]

Where:
- `z` = standardized value  

---

## 📊 Z-Score Formula

\[
z = \frac{x - \mu}{\sigma}
\]

Where:
- `x` = actual value  
- `μ` = mean  
- `σ` = standard deviation  

---

## 🎯 Example

A student scored **85 marks** in an exam:
- Mean (μ) = 70  
- Standard deviation (σ) = 10  

\[
z = \frac{85 - 70}{10} = 1.5
\]

👉 This means the score is **1.5 standard deviations above the mean**.
