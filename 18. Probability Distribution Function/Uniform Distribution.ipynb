# Uniform Distribution

## 📌 Introduction
Uniform Distribution is a type of probability distribution where **all outcomes are equally likely** within a given range.

👉 In simple terms:  
Every value has the same probability of occurring.

---

## 📊 Types of Uniform Distribution

### 1. Discrete Uniform Distribution
- Deals with **finite/countable values**
- Each value has equal probability

#### 📍 Example:
Rolling a fair die:
P(1) = P(2) = P(3) = P(4) = P(5) = P(6) = 1/6

#### 📐 Formula:
If there are `n` possible outcomes:

P(X = x) = 1 / n


---

### 2. Continuous Uniform Distribution
- Deals with **continuous range of values**
- Probability is spread evenly over an interval [a, b]

#### 📍 Example:
Random number between 0 and 1

#### 📐 Probability Density Function (PDF):

f(x) = 1 / (b - a), for a ≤ x ≤ b
0, otherwise


---

## 📉 Graph Representation

### Discrete Uniform:
- Equal height bars for each value

### Continuous Uniform:
- Rectangular (flat) graph between `a` and `b`

---

## 📌 Key Properties

### Discrete Uniform:
- Mean:

μ = (n + 1) / 2 (for values 1 to n)


- Variance:

σ² = (n² - 1) / 12


---

### Continuous Uniform:
- Mean:

μ = (a + b) / 2


- Variance:

σ² = (b - a)² / 12


---

## ⚡ Characteristics
- All outcomes are equally likely
- Simple and easy to understand
- No bias toward any value
- Used when information is limited

---

## 📍 Real-Life Examples
- Rolling a fair die (Discrete)
- Lottery number selection (Discrete)
- Random number generation (Continuous)
- Waiting time within a fixed interval (Continuous)

---

## 🆚 Discrete vs Continuous Uniform

| Feature              | Discrete Uniform         | Continuous Uniform       |
|---------------------|--------------------------|--------------------------|
| Data Type           | Countable values         | Continuous range         |
| Example             | Dice roll                | Random number [0,1]      |
| Probability         | Exact value probability  | Interval probability     |
| Graph               | Bar graph                | Rectangular curve        |

---

## ✅ Summary
- Uniform Distribution means **equal probability for all outcomes**
- Two types:
  - Discrete Uniform Distribution
  - Continuous Uniform Distribution
- Widely used in simulations, modeling, and probability theory
