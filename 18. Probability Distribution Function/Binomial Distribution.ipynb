# 📊 Binomial Distribution

## 📌 Overview

The **Binomial Distribution** is a discrete probability distribution that models the number of successes in a fixed number of independent trials of a binary (yes/no) experiment.

Each trial has only **two possible outcomes**:

- Success (1)
- Failure (0)

---

## 🧠 Key Characteristics

- Fixed number of trials → `n`
- Each trial is independent
- Probability of success remains constant → `p`
- Only two outcomes per trial (Bernoulli trial)

---

## 📐 Probability Mass Function (PMF)

The probability of getting exactly `k` successes in `n` trials is given by:

\[
P(X = k) = \binom{n}{k} p^k (1 - p)^{n - k}
\]

Where:

- `n` = number of trials
- `k` = number of successes
- `p` = probability of success
- `1 - p` = probability of failure
- \(\binom{n}{k}\) = combinations = `n! / (k!(n-k)!)`

---

## 📊 Mean and Variance

- **Mean (Expected Value):**
  \[
  \mu = n \cdot p
  \]

- **Variance:**
  \[
  \sigma^2 = n \cdot p \cdot (1 - p)
  \]

- **Standard Deviation:**
  \[
  \sigma = \sqrt{n \cdot p \cdot (1 - p)}
  \]

---

## 🎯 Example

### Problem:

A fair coin is tossed **10 times**. What is the probability of getting exactly **6 heads**?

### Solution:

- `n = 10`
- `k = 6`
- `p = 0.5`

\[
P(X = 6) = \binom{10}{6} (0.5)^6 (0.5)^4
\]

\[
P(X = 6) = 210 \times (0.5)^{10} = 210 / 1024 \approx 0.205
\]

---

## 🧪 Real-Life Applications

- 🎲 Number of heads in coin tosses
- 📦 Defective items in a batch
- 📧 Number of emails opened
- 🏏 Number of successful shots in cricket
- 💻 Success/failure of network requests

---

## ⚙️ Python Implementation

```python
from scipy.stats import binom

# Parameters
n = 10  # trials
p = 0.5 # probability of success
k = 6   # successes

# PMF
probability = binom.pmf(k, n, p)

print("Probability:", probability)
```


````python
import matplotlib.pyplot as plt
from scipy.stats import binom

n = 10
p = 0.5

x = list(range(n + 1))
y = [binom.pmf(i, n, p) for i in x]

plt.bar(x, y)
plt.xlabel("Number of Successes")
plt.ylabel("Probability")
plt.title("Binomial Distribution")
plt.show()
````
