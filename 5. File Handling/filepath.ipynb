```python
# Working with the file paths

# create new directory

import os

new_dir = "new_directory"
os.mkdir(new_dir)
# create new file in the new directory
new_file_path = os.path.join(new_dir, "new_file.txt")
with open(new_file_path, "w") as f:
    f.write("This is a new file.")
# list files in the new directory
files = os.listdir(new_dir)
```

```python
# Listing the files and directories in the current directory

items= os.listdir(".")
print(items)
```

```python
# Joining paths

dir_name = "my_directory"
file_name = "my_file.txt"
full_path = os.path.join(dir_name, file_name)
print(full_path)
```

```python
# check the existence of a file or directory

path = "my_directory/my_file.txt"
if os.path.exists(path):
    print(f"{path} exists.")
else:
    print(f"{path} does not exist.")
```

```python
# CHECKING IF A PATH IS A FILE OR DIRECTORY

path = "./new_directory/new_file.txt"
if os.path.isfile(path):
    print(f"{path} is a file.") 
elif os.path.isdir(path):
    print(f"{path} is a directory.")
else:
    print(f"{path} does not exist.")
```

```python
# Getting absolute path

relative_path = "new_directory/new_file.txt"
absolute_path = os.path.abspath(relative_path)
print(f"Relative path: {relative_path}")
print(f"Absolute path: {absolute_path}")
```
