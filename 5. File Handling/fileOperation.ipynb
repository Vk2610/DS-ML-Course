# File Operation - Read and Write Files in Python

File handling is a crucial aspect of programming, allowing you to read from and write to files on your system. In Python, you can easily perform file operations using built-in functions. Below are some common file operations:
## 1. Opening a FileTo open a file, you can use the `open()` function. It takes two arguments: the file name and the mode (e.g., 'r' for reading, 'w' for writing, 'a' for appending).
```python# Open a file for reading
file = open('example.txt', 'r')
```

## 2. Reading from a FileTo read the contents of a file, you can use methods like `read()`, `readline()`, or `readlines()`.
```python# Read the entire content of the file  
content = file.read()
print(content)
```

## 3. Writing to a FileTo write to a file, you can open it in write mode ('w') or append mode ('a'). Use the `write()` method to add content to the file.
```python# Open a file for writing  
file = open('example.txt', 'w')
# Write some content to the file
file.write('Hello, World!')
# Close the file after writing
file.close()
```
## 4. Closing a FileIt's important to close a file after you're done with it to free up system resources. You can use the `close()` method for this.
```python# Close the file after reading or writing
file.close()
```

````python
# Read a whole file 

with open('file.txt','r') as f:
    data = f.read()
    print(data)
````

````python
# Read file line by line

with open('file.txt','r') as f:
    for line in f:
        print(line.strip()) # strip() is used to remove the newline character at the end of each line
````

````python
# Open a file in write mode and write some data to it

with open('file.txt','w') as f:
    f.write('Hello, this is a new file.\n')
    f.write('This file is created using Python.\n')
    
print('Data written to file successfully.')
````

````python
# Open modified file in read mode 
with open('file.txt','r') as f:
    data = f.read()
    print(data)
````

Write file function overwrites the existing content of the file. If you want to add content without overwriting, use append mode ('a') instead of write mode ('w').
```python# Open a file for appending

````python
# Modify file content using append mode
with open('file.txt','a') as f:
    f.write('This line is added using append mode.\n')
# Read the modified file to see the changes
with open('file.txt','r') as f:
    data = f.read()
    print(data)
````

````python
# Writing list of lines to a file
lines = ['Line 1\n', 'Line 2\n', 'Line 3\n']
with open('file.txt','w') as f:
    f.writelines(lines)
# Read the file to see the changes
with open('file.txt','r') as f:
    data = f.read()
    print(data)
````

````python
### Working with Binary Files

data = b'\x00\x01\x02\x03\x04\x05\x06\x07\x08\x09'

with open('binary_file.bin','wb') as f:
    f.write(data)
with open('binary_file.bin','rb') as f:
    binary_data = f.read()
    print(binary_data)
````

````python
### Read the content from a source file and write to a destination text file 
# Copying a text file 

with open('file.txt','r') as source_file:
    content = source_file.read()
    
with open('copy_file.txt','w') as dest_file:
    dest_file.write(content)  
````

W+ mode allows you to read and write to the file. If the file does not exist, it will be created. If it already exists, the content will be overwritten.
```python# Open a file for reading and writing

````python
# Writing and then reading a file

with open('file.txt','w+') as f:
    f.write('This file is opened in write and read mode.\n')
    f.write('You can write and read from the same file.\n')
    f.seek(0) # Move the cursor to the beginning of the file
    data = f.read()
    print(data)
````
