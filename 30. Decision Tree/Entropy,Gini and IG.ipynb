# 📊 Entropy, Gini Impurity, and Information Gain

## 📌 Introduction

When building a Decision Tree, the most important task is deciding **which feature should be used to split the data at each node**.

To make this decision, Decision Trees use measures of impurity and information:

1. **Entropy** – Measures uncertainty or randomness in data.
2. **Gini Impurity (Gini Index)** – Measures how often a randomly chosen element would be incorrectly classified.
3. **Information Gain** – Measures the reduction in uncertainty after splitting the dataset.

These metrics help the Decision Tree choose the best feature for creating branches.

---

# 🎯 Why Are These Metrics Needed?

Consider a dataset for email classification:

| Email | Spam |
|---------|---------|
| Offer | Yes |
| Win Prize | Yes |
| Meeting Schedule | No |
| Project Update | No |

Before splitting the dataset, the classes are mixed.

A Decision Tree needs to find a feature that creates the purest possible groups.

For example:

```
Before Split

Spam = 2
Not Spam = 2
```

After splitting:

```
Group 1:
Spam = 2
Not Spam = 0

Group 2:
Spam = 0
Not Spam = 2
```

This split is perfect because each group contains only one class.

Entropy, Gini Impurity, and Information Gain help identify such splits.

---

# 🌟 Entropy

## Definition

Entropy is a measure of uncertainty, randomness, or disorder in a dataset.

A node is considered:

- Pure if all samples belong to one class.
- Impure if samples belong to multiple classes.

Entropy quantifies this impurity.

---

## Formula

\[
Entropy(S) = -\sum_{i=1}^{n} p_i \log_2(p_i)
\]

Where:

- \(S\) = Dataset
- \(p_i\) = Probability of class \(i\)

---

# 📊 Entropy Interpretation

| Entropy Value | Meaning |
|--------------|---------|
| 0 | Completely Pure |
| 0.5 | Moderately Mixed |
| 1 | Maximum Uncertainty (Binary Classification) |

# 🌟 Gini Impurity (Gini Index)

## Definition

Gini Impurity measures the probability of incorrectly classifying a randomly selected element if it is labeled according to the class distribution in the dataset.

It is used by the CART (Classification and Regression Tree) algorithm.

---

## Formula

\[
Gini = 1 - \sum_{i=1}^{n} p_i^2
\]

Where:

- \(p_i\) = Probability of class \(i\)

---

# 📊 Gini Interpretation

| Gini Value | Meaning |
|------------|----------|
| 0 | Pure Node |
| 0.5 | Highly Impure (Binary Classification) |

---

| Yes |

\[
P(Yes)=1
\]

\[
1 - (1^2)
\]

\[
=0
\]

```
Gini = 0
```

Perfectly pure node.

---

# Entropy vs Gini Impurity

| Feature | Entropy | Gini Impurity |
|----------|----------|--------------|
| Formula | Uses Logarithms | Uses Squares |
| Range | 0 – 1 | 0 – 0.5 |
| Computation | Slower | Faster |
| Used In | ID3, C4.5 | CART |
| Complexity | Higher | Lower |

# 🌟 Information Gain

## Definition

Information Gain measures how much uncertainty is reduced after splitting the dataset.

The Decision Tree selects the feature with the highest Information Gain.

---

## Formula

\[
Information\ Gain = Entropy(Parent) - Weighted\ Entropy(Children)
\]

---

# Why Information Gain?

Suppose the parent node has:

```
Entropy = 1
```

After splitting:

```
Child 1 Entropy = 0
Child 2 Entropy = 0
```

The split completely removes uncertainty.

Therefore:

```
Information Gain = 1
```

Perfect split.

# How Decision Trees Use These Metrics

```
        |

```

# 📚 Summary

- **Entropy** measures randomness or uncertainty in a dataset.
- **Gini Impurity** measures the likelihood of incorrect classification.
- **Information Gain** measures the reduction in uncertainty after a split.
- Decision Trees use these metrics to determine the best feature for splitting data.
- Lower Entropy and Gini values indicate purer nodes.
- Higher Information Gain indicates a better split.

These concepts form the mathematical foundation of Decision Tree learning and are essential for understanding advanced tree-based algorithms such as Random Forest and XGBoost.

````python
# Implementation

import pandas as pd
import numpy as np
from math import log2

# Sample Dataset
data = {
    "Weather": ["Sunny", "Sunny", "Rainy", "Rainy", "Sunny", "Rainy"],
    "Windy": ["No", "Yes", "No", "Yes", "No", "No"],
    "Play": ["No", "No", "Yes", "Yes", "Yes", "Yes"]
}

df = pd.DataFrame(data)

print("Dataset:")
print(df)
print("\n")


# --------------------------
# Entropy Calculation
# --------------------------
def entropy(target_column):
    values, counts = np.unique(target_column, return_counts=True)

    ent = 0

    for count in counts:
        probability = count / len(target_column)

        ent -= probability * log2(probability)

    return ent


# --------------------------
# Gini Impurity Calculation
# --------------------------
def gini_impurity(target_column):
    values, counts = np.unique(target_column, return_counts=True)

    gini = 1

    for count in counts:
        probability = count / len(target_column)

        gini -= probability ** 2

    return gini


# --------------------------
# Information Gain
# --------------------------
def information_gain(data, feature, target):

    parent_entropy = entropy(data[target])

    values, counts = np.unique(data[feature], return_counts=True)

    weighted_entropy = 0

    for value, count in zip(values, counts):

        subset = data[data[feature] == value]

        subset_entropy = entropy(subset[target])

        weighted_entropy += (count / len(data)) * subset_entropy

    gain = parent_entropy - weighted_entropy

    return gain


# --------------------------
# Parent Node Metrics
# --------------------------
print("Parent Node Metrics")
print("--------------------")

parent_entropy = entropy(df["Play"])
parent_gini = gini_impurity(df["Play"])

print(f"Entropy       : {parent_entropy:.4f}")
print(f"Gini Impurity : {parent_gini:.4f}")

print("\n")


# --------------------------
# Information Gain for Features
# --------------------------
print("Feature Analysis")
print("--------------------")

features = ["Weather", "Windy"]

best_feature = None
max_gain = -1

for feature in features:

    gain = information_gain(df, feature, "Play")

    print(f"{feature} -> Information Gain = {gain:.4f}")

    if gain > max_gain:
        max_gain = gain
        best_feature = feature

print("\n")


# --------------------------
# Best Feature Selection
# --------------------------
print("Best Feature for Split")
print("----------------------")

print(f"Feature: {best_feature}")
print(f"Information Gain: {max_gain:.4f}")
````
