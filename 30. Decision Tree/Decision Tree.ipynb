# 🌳 Decision Tree Algorithm

## 📌 Introduction

A **Decision Tree** is a supervised machine learning algorithm used for both **classification** and **regression** tasks. It works by creating a tree-like structure where each internal node represents a decision based on a feature, each branch represents an outcome of that decision, and each leaf node represents the final prediction.

Decision Trees are one of the most intuitive machine learning algorithms because they mimic human decision-making processes.

---

## 🎯 Real-Life Example

Imagine deciding whether to play cricket based on weather conditions.

```

                Weather?
               /        \
           Sunny       Rainy
           /              \
      Humidity?        Don't Play
      /      \
   High      Low
   /           \
Don't Play    Play

```

The tree asks a series of questions and reaches a final decision.

---

# 📖 How Decision Trees Work

A Decision Tree divides the dataset into smaller subsets while simultaneously developing an associated tree structure.

The process involves:

1. Selecting the best feature.
2. Splitting the dataset based on that feature.
3. Repeating the process recursively.
4. Stopping when:
   - All data belongs to one class.
   - Maximum depth is reached.
   - No further useful split exists.

---

# 🌲 Structure of a Decision Tree

```

                Root Node
                     |
            ----------------
            |              |
      Internal Node   Internal Node
            |              |
         Branches       Branches
            |              |
         Leaf Node     Leaf Node

```

### Components

### 1. Root Node

- Starting point of the tree.
- Represents the entire dataset.

### 2. Internal Node

- Represents a feature/attribute.
- Performs a test on data.

### 3. Branch

- Outcome of a test.

### 4. Leaf Node

- Final prediction or decision.

---

# ⚙️ Decision Tree Algorithm Steps

### Step 1: Start with Entire Dataset

Consider:

| Outlook | Temperature | Play |
|----------|------------|--------|
| Sunny | Hot | No |
| Sunny | Mild | No |
| Rainy | Cool | Yes |
| Rainy | Mild | Yes |

---

### Step 2: Calculate Impurity

The algorithm determines which feature best separates classes.

Common impurity measures:

- Entropy
- Information Gain
- Gini Index

---

### Step 3: Select Best Feature

Feature with highest Information Gain or lowest Gini Impurity is selected.

---

### Step 4: Split Dataset

Create branches based on feature values.

---

### Step 5: Repeat Recursively

Continue splitting until stopping criteria are met.

---

# 📊 Entropy

Entropy measures randomness or impurity in data.

## Formula

\[
Entropy(S) = - \sum p_i \log_2(p_i)
\]

Where:

- \(p_i\) = probability of class i

---

### Example

Dataset:

| Class | Count |
|---------|-------|
| Yes | 8 |
| No | 8 |

Probability:

\[
P(Yes)=0.5
\]

\[
P(No)=0.5
\]

Entropy:

\[
-(0.5\log_2 0.5 + 0.5\log_2 0.5)
\]

\[
=1
\]

Maximum entropy = 1 (high uncertainty)

---

## Entropy Interpretation

| Entropy | Meaning |
|----------|---------|
| 0 | Pure node |
| 1 | Maximum uncertainty |

---

# 📈 Information Gain

Information Gain tells how much uncertainty decreases after splitting.

## Formula

\[
IG = Entropy(Parent) - Weighted\ Entropy(Children)
\]

Higher Information Gain means better split.

---

### Example

Parent Entropy:

\[
1
\]

After split:

\[
0.3
\]

Information Gain:

\[
1 - 0.3 = 0.7
\]

This is a good split.

---

# 📊 Gini Index

Another measure of impurity used by CART Decision Trees.

## Formula

\[
Gini = 1 - \sum p_i^2
\]

---

### Example

| Class | Probability |
|---------|------------|
| Yes | 0.5 |
| No | 0.5 |

\[
Gini = 1 - (0.5^2 + 0.5^2)
\]

\[
= 1 - (0.25+0.25)
\]

\[
=0.5
\]

---

## Gini Interpretation

| Gini Value | Meaning |
|-------------|----------|
| 0 | Pure node |
| 0.5 | Highly mixed |

# 🧮 Example Using Scikit-Learn

## Dataset

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split

X = [[1,25],
     [0,30],
     [1,35],
     [0,40]]

y = [0,1,0,1]
```

---

## Training Model

```python
model = DecisionTreeClassifier()

model.fit(X, y)
```

---

## Prediction

```python
prediction = model.predict([[1,28]])

print(prediction)
```

Output:

```python
[0]
```

---

# 🌳 Visualizing Decision Tree

```python
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt

plot_tree(model, filled=True)

plt.show()
```

# ⚡ Important Hyperparameters

## max_depth

Maximum depth of tree.

```python
DecisionTreeClassifier(max_depth=5)
```

---

## min_samples_split

Minimum samples required to split a node.

```python
DecisionTreeClassifier(min_samples_split=4)
```

---

## min_samples_leaf

Minimum samples required in a leaf node.

```python
DecisionTreeClassifier(min_samples_leaf=2)
```

---

## criterion

Split quality measure.

```python
DecisionTreeClassifier(criterion="gini")
```

or

```python
DecisionTreeClassifier(criterion="entropy")
```

# 🔥 Overfitting in Decision Trees

When a tree memorizes training data instead of learning patterns, it overfits.

Example:

```

Training Accuracy = 100%
Testing Accuracy = 65%

```

---

## Solutions

### Pruning

Removing unnecessary branches.

### Limit Tree Depth

```python
DecisionTreeClassifier(max_depth=5)
```

### Increase Minimum Samples

```python
DecisionTreeClassifier(min_samples_leaf=5)
```

---

# ✂️ Pruning

Pruning reduces tree complexity.

## Pre-Pruning

Stop tree growth early.

Methods:

- max_depth
- min_samples_split
- min_samples_leaf

---

## Post-Pruning

Build full tree first and then remove branches.

Benefits:

- Better generalization
- Reduced overfitting
