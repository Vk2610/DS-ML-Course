# Numpy

Numpy is a fundamental library for scientific computing in python. It provides support for arrays and matrices, along with a collection of mathematical functions to operate on these data structures 

```python
pip install numpy
```

```python
import numpy as np 

# create arrays using numpy

# 1D array
a = np.array([1,2,3,4,5])
print(a)
print(type(a))
print(a.shape)
```

```python
# reshape array 

a2 = a.reshape(1,5) # 1 row and 5 columns
print(a2)
print(a2.shape)
print(type(a2))
```

```python
a3 = np.array([[1,2,3],[4,5,6]])
print(a3)
print(type(a3))
print(a3.shape)
```

```python
np.arange(0,10,2)
```

```python
np.ones((3,4))
```

```python
# identity matrix

np.eye(4)
```

```python
arr = np.array([[1,2,3],[4,5,6]])

print("Array:\n", arr)
print("Shape:", arr.shape)
print("Number of dimensions:", arr.ndim)
print("Size:", arr.size)
print("Data type:", arr.dtype)
print("Data type name:", arr.dtype.name)
print("Item size in bytes:", arr.itemsize)
```
