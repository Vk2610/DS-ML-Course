# Pandas - DataFrames and Series

Pandas is a powerful library in Python for data manipulation and analysis. It provides two main data structures: Series and DataFrame. A Series is a one-dimensional array-like object that can hold any data type, while a DataFrame is a two-dimensional table with labeled axes (rows and columns).

```python
# Series
# A Series is a one-dimensional array-like object that can hold any data type. It has an index that labels each element in the Series.

# Create a Series from a list
s = pd.Series([10, 20, 30, 40, 50])
print(s)
```
