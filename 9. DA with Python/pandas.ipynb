# Pandas - DataFrames and Series

Pandas is a powerful library in Python for data manipulation and analysis. It provides two main data structures: Series and DataFrame. A Series is a one-dimensional array-like object that can hold any data type, while a DataFrame is a two-dimensional table with labeled axes (rows and columns).

```python
import pandas as pd
```

```python
# Series
# A Series is a one-dimensional array-like object that can hold any data type. It has an index that labels each element in the Series.

# Create a Series from a list
s = pd.Series([10, 20, 30, 40, 50])
print(s)
```

```python
# create a Series from a dictionary

data = {'a': 1, 'b': 2, 'c': 3}
s = pd.Series(data)
print(s)
```

```python
data = [1, 2, 3, 4, 5]
index = ['a', 'b', 'c', 'd', 'e']
s = pd.Series(data, index=index)
print(s)
```

```python
# DataFrame
# A DataFrame is a two-dimensional tabular data structure with labeled axes (rows and columns

# Create a DataFrame from a dictionary of lists
data = {'Name': ['Alice', 'Bob', 'Charlie'],
        'Age': [25, 30, 35],
        'City': ['New York', 'Los Angeles', 'Chicago']}
df = pd.DataFrame(data)
print(df)
```

```python
# Create a dataframe from a list of dictionaries

data = [{'Name': 'Alice', 'Age': 25, 'City': 'New York'},
        {'Name': 'Bob', 'Age': 30, 'City': 'Los Angeles'},
        {'Name': 'Charlie', 'Age': 35, 'City': 'Chicago'}]
df = pd.DataFrame(data)
print(df)
```

```python
# Accessing specified element

df.at[1, 'Product']  # Access the 'Product' value in the second row
```

```python
# Accessing specified element using iat

df.iat[1, 3]  # Access the element in the second row and first column
```

```python
# Data manipulation with dataframes

# add new column to the dataframe
df['Total'] = df['Quantity'] * df['UnitPrice']
df
```

```python
# Remove a column from the dataframe
df.drop('Total', axis=1, inplace=True)  # drop - is temporary, inplace - is permanent
df
```

```python
# Add a +1 to the 'Quantity' column
df['Quantity'] = df['Quantity'] + 1
df
```

```python
# Recalculate the 'Total' column after updating 'Quantity'
df['TotalAmount'] = df['Quantity'] * df['UnitPrice']
df
```
