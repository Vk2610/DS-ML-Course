# Pickle file 
- Pickle in python is primarily used for serializing and deserializing Python objects.

- It is the process of converting python object into a byte stream, and the process of converting byte stream back to python object is called unpickling.

```python
# convert xlsx to pickle file

df.to_pickle('employee.pkl')
```

```python
pd.read_pickle('employee.pkl')
```
