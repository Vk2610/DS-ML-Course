# 📘 Ridge Regression in Machine Learning


## 📖 Introduction
Ridge Regression extends Linear Regression using L2 regularization to reduce overfitting and handle multicollinearity.

## ❓ What is Ridge Regression?
y = b0 + b1x1 + ... + bnxn

Cost:
RSS + λ Σ bj²

## 🎚 Effect of Lambda
λ=0 no regularization  
Large λ stronger shrinkage

## ✅ Advantages
- Reduces overfitting
- Stabilizes coefficients

## ❌ Disadvantages
- No feature elimination
- Requires tuning λ

## ⚖️ Ridge vs Lasso
Ridge uses L2.
Lasso uses L1.


```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression,Ridge
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import PolynomialFeatures

np.random.seed(0)
X=np.sort(np.random.rand(30)).reshape(-1,1)
y=np.sin(2*np.pi*X).ravel()+np.random.randn(30)*0.2
X_plot=np.linspace(0,1,200).reshape(-1,1)
```

## Graph: Linear vs Ridge

```python
linear=make_pipeline(
PolynomialFeatures(10),
LinearRegression()
)
ridge=make_pipeline(
PolynomialFeatures(10),
Ridge(alpha=1.0)
)

linear.fit(X,y)
ridge.fit(X,y)

plt.figure()
plt.scatter(X,y)
plt.plot(X_plot,linear.predict(X_plot),label='Linear')
plt.plot(X_plot,ridge.predict(X_plot),label='Ridge')
plt.legend()
plt.title("Linear vs Ridge")
plt.show()
```

## Graph: Lambda Effect

```python
alphas=[0.01,1,100]

plt.figure()
plt.scatter(X,y)

for a in alphas:
    model=make_pipeline(
    PolynomialFeatures(10),
    Ridge(alpha=a)
    )
    model.fit(X,y)
    plt.plot(X_plot,model.predict(X_plot),label=f'alpha={a}')

plt.legend()
plt.title("Effect of Lambda")
plt.show()
```

## Python Implementation

```python
X2=np.array([[1],[2],[3],[4],[5]])
y2=np.array([2,4,5,4,5])

model=Ridge(alpha=1.0)
model.fit(X2,y2)

print(model.coef_)
print(model.intercept_)
```
