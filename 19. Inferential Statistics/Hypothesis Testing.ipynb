
# Hypothesis Testing & Its Mechanism

## 📌 Introduction
**Hypothesis Testing** is a statistical method used to make decisions or inferences about a population based on sample data.

👉 In simple terms:  
It helps us determine whether there is **enough evidence** to support a claim.

---

## 📊 Key Concepts

### 1. Null Hypothesis (H₀)
- Default assumption
- States **no effect or no difference**

👉 Example:  
H₀: The average marks = 50

---

### 2. Alternative Hypothesis (H₁ / Hₐ)
- Opposes the null hypothesis
- Indicates **presence of effect or difference**

👉 Example:  
H₁: The average marks ≠ 50

---

### 3. Significance Level (α)
- Probability of rejecting H₀ when it is true
- Common values: **0.05, 0.01**

---

### 4. Test Statistic
- A value calculated from sample data
- Used to decide whether to reject H₀

---

### 5. P-value
- Probability of obtaining results at least as extreme as observed
- Helps in decision making

---

## ⚙️ Mechanism of Hypothesis Testing (Step-by-Step)

### Step 1: State the Hypotheses
- Define **H₀** (Null Hypothesis)
- Define **H₁** (Alternative Hypothesis)

---

### Step 2: Choose Significance Level (α)
- Usually **0.05 (5%)**

---

### Step 3: Select the Appropriate Test
Depending on data and situation:
- Z-test  
- T-test  
- Chi-square test  
- ANOVA  

---

### Step 4: Calculate Test Statistic
- Use sample data to compute the statistic

---

### Step 5: Determine P-value or Critical Value
- Compare test statistic with critical value  
**OR**  
- Calculate p-value

---

### Step 6: Make Decision
- If **p-value ≤ α** → Reject H₀  
- If **p-value > α** → Fail to reject H₀  

---

### Step 7: Conclusion
- Interpret the result in context of the problem

---

## 📉 Types of Hypothesis Tests

### 1. One-Tailed Test
- Tests in **one direction only**
- Example: Mean > 50

---

### 2. Two-Tailed Test
- Tests in **both directions**
- Example: Mean ≠ 50

---

## ⚠️ Errors in Hypothesis Testing

### Type I Error (False Positive)
- Rejecting H₀ when it is true  
- Probability = α  

---

### Type II Error (False Negative)
- Failing to reject H₀ when it is false  
- Probability = β  

---

## 📍 Real-Life Examples
- Testing effectiveness of a new drug  
- Checking if a coin is fair  
- Quality control in manufacturing  
- A/B testing in websites  

---

## 🆚 P-value vs Significance Level

| Feature        | P-value                     | Significance Level (α)   |
|----------------|----------------------------|--------------------------|
| Definition     | Computed probability       | Predefined threshold     |
| Role           | Evidence against H₀        | Decision boundary        |
| Decision Rule  | Compare with α             | Used for comparison      |

---

## ✅ Summary
- Hypothesis testing helps in **decision making using data**
- Based on **null and alternative hypotheses**
- Uses **test statistics and p-values**
- Final decision: **Reject or Fail to Reject H₀**
