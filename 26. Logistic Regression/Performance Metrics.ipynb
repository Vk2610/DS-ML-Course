# 📊 Performance Metrics in Machine Learning

This notebook covers important evaluation metrics used in Machine Learning for both classification and regression problems.

---

## 📌 1. Classification Metrics

### Confusion Matrix

|               | Predicted Positive | Predicted Negative |
|--------------|-------------------|-------------------|
| Actual Positive | TP | FN |
| Actual Negative | FP | TN |

- TP: True Positive
- TN: True Negative
- FP: False Positive
- FN: False Negative


```python
from sklearn.metrics import confusion_matrix

y_true = [1, 0, 1, 1, 0, 1]
y_pred = [1, 0, 0, 1, 0, 1]

cm = confusion_matrix(y_true, y_pred)
cm
```

### Accuracy

$$Accuracy = \frac{TP + TN}{TP + TN + FP + FN}$$

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_true, y_pred)
```

### Precision

$$Precision = \frac{TP}{TP + FP}$$

```python
from sklearn.metrics import precision_score
precision_score(y_true, y_pred)
```

### Recall

$$Recall = \frac{TP}{TP + FN}$$

```python
from sklearn.metrics import recall_score
recall_score(y_true, y_pred)
```

### F1 Score

$$F1 = 2 \cdot \frac{Precision \cdot Recall}{Precision + Recall}$$

```python
from sklearn.metrics import f1_score
f1_score(y_true, y_pred)
```

### ROC-AUC Curve

ROC curve shows trade-off between TPR and FPR.


```python
from sklearn.metrics import roc_curve, auc
import matplotlib.pyplot as plt

y_scores = [0.9, 0.1, 0.4, 0.8, 0.2, 0.7]

fpr, tpr, _ = roc_curve(y_true, y_scores)
roc_auc = auc(fpr, tpr)

plt.plot(fpr, tpr)
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC Curve')
plt.show()

roc_auc
```

## 📌 2. Regression Metrics


### Mean Absolute Error (MAE)

$$MAE = \frac{1}{n} \sum |y - \hat{y}|$$

```python
from sklearn.metrics import mean_absolute_error

y_true_reg = [3, -0.5, 2, 7]
y_pred_reg = [2.5, 0.0, 2, 8]

mean_absolute_error(y_true_reg, y_pred_reg)
```

### Mean Squared Error (MSE)

$$MSE = \frac{1}{n} \sum (y - \hat{y})^2$$

```python
from sklearn.metrics import mean_squared_error
mean_squared_error(y_true_reg, y_pred_reg)
```

### Root Mean Squared Error (RMSE)

$$RMSE = \sqrt{MSE}$$

```python
import numpy as np
np.sqrt(mean_squared_error(y_true_reg, y_pred_reg))
```

### R² Score (Coefficient of Determination)

$$R^2 = 1 - \frac{SS_{res}}{SS_{tot}}$$

```python
from sklearn.metrics import r2_score
r2_score(y_true_reg, y_pred_reg)
```
