# Importing modules in python : Modules and packages 

In Python, moduls and packages help organize and reuse code. Here's a comprehensive guide on hoe to import packages

```python
import math 
print(math.sqrt(16))
```

```python
from math import sqrt,pi

print(sqrt(25))
print(pi)
```

```python
import numpy as np # allias for numpy
print(np.sqrt(16))
```

```python
import numpy as np 
np.array([1, 2, 3, 4, 5])
```

```python
from math import * #importing all functions from math module
print(sqrt(36))
print(pi)
print(cos(0))
print(sin(40))
```

```python
# Custom package import
from package.maths import addition, subtraction

print(addition(5, 3))
print(subtraction(5, 3))
```

```python
# another way to import custom package
from package import maths
print(maths.addition(10, 5))
print(maths.subtraction(10, 5))
```

```python
from package.subpackages.mult import mult
print(mult(4, 5))
```
