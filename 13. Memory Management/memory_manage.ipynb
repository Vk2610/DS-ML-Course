## Memory Management Best Practices

1. Use Local Variables: Local variables are automatically deallocated when they go out of scope, which can help reduce memory usage.

2. Avoid Circular References: Be cautious when creating circular references, as they can lead to memory leaks. Use weak references if necessary.

3. Use Generators: Generators can help reduce memory usage by generating items on-the-fly instead of storing them all in memory at once.

4. Explicitly Delete Unused Objects: Use the `del` statement to explicitly delete objects that are no longer needed, especially if they consume a lot of memory.

5. Profile Memory Usage: Use tools like `memory_profiler` or `tracemalloc` to profile your code's memory usage and identify potential memory leaks.

```python
import gc 

class MyObject:
    def __init__(self,name):
      self.name = name
      print(f"{self.name} is created")
      
    def __del__(self):
      print(f"{self.name} is destroyed")
      
# create circular reference

obj1 = MyObject("Object 1")
obj2 = MyObject("Object 2")
obj1.ref = obj2
obj2.ref = obj1

del obj1
del obj2

# manually trigger the garbage collection

gc.collect()
```

```python
# Generators for memory efficiency
# Generators allow you to produce items one at a time, which can be more memory efficient than creating a list of all items at once.

def generate_numbers(n):
  for i in range(n):
    yield i
    
# using the generator
for number in generate_numbers(10):
  print(number)
  if number > 10:
    break
```

```python
# Profiling memory using with tracemalloc

import tracemalloc

def create_list():
  return [i for i in range(1000000)]

def main():
  tracemalloc.start()
  
  create_list()
  
  current, peak = tracemalloc.get_traced_memory()
  print(f"Current memory usage: {current / 10**6} MB")
  print(f"Peak memory usage: {peak / 10**6} MB")
  
  tracemalloc.stop()
  
main()
```
