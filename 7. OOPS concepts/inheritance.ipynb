# Inheritance in python 
Inheritance is a fundamental object-oriented programming concept that allows a new class (called a child or subclass) to inherit attributes and behaviors (methods) from an existing class (called a parent or superclass). This promotes code reusability and establishes a natural hierarchical relationship between classes.

```python
# Inheritance (single inheritance)
# Parent class
class Car:
  def __init__ (self,windows,doors,enginetype):
    self.windows = windows
    self.doors = doors
    self.enginetype = enginetype
  def drive(self):
    print("The car is driving")
```

```python
car1=Car(4,4,"Petrol")
print(car1.windows)
print(car1.doors)
print(car1.enginetype)
car1.drive()
```

```python
class Tesla(Car):
  def __init__(self,windows,doors,enginetype,is_Self_driving):
    super().__init__(windows,doors,enginetype)
    self.is_Self_driving = is_Self_driving
  def autopilot(self):
    print("The car is in autopilot mode")
    
tesla1 = Tesla(4,4,"Electric",True)
print(tesla1.windows)
print(tesla1.doors)
print(tesla1.enginetype)
print(tesla1.is_Self_driving)
tesla1.drive()
tesla1.autopilot()
```

```python
# Multiple inheritance 
#  -- when a class inherits from more than one parent class, it is called multiple inheritance. The child class can access the attributes and methods of all the parent classes.

class Animal:
  def __init__(self,name):
    self.name = name
  def eat(self):
    print(f"{self.name} is eating")
    
# Parent class 2
class Pet:
  def __init__(self,owner):
    self.owner = owner
  def play(self):
    print(f"{self.owner}'s pet is playing")
    
# Child class
class Dog(Animal,Pet):
  def __init__(self,name,owner,breed):
    Animal.__init__(self,name)
    Pet.__init__(self,owner)
    self.breed = breed
  def bark(self):
    print(f"{self.name} is barking")
    
dog1 = Dog("Buddy","Alice","Golden Retriever")
print(dog1.name)  
print(dog1.owner)
print(dog1.breed)
dog1.eat()
dog1.play()
dog1.bark()
```
