# Magic Methods

Magic methods in python, also known as dunder methods (double underscore methods), are special methods that start and end with double underscores. These methods enable you to define the behavior of objects for built-in operations such as arithmetic operations, comparisons and more

**Magic methods in Python** (also called **dunder methods**, because they look like `__method__`) are **special methods** that let you define how your objects behave with built-in Python operations like `+`, `len()`, `print()`, comparisons, iteration, etc.

They allow **operator overloading** and **custom behavior** for classes.

---

## 🔹 What are Magic Methods?

Magic methods are **automatically called by Python**, not usually called directly by you.

Example:

```python
len(obj)      # calls obj.__len__()
obj + other   # calls obj.__add__(other)
print(obj)    # calls obj.__str__()
```

---

## 🔹 Commonly Used Magic Methods

### 1️⃣ Object Initialization & Representation

```python
__init__()   # constructor
__str__()    # user-friendly string
__repr__()   # developer-friendly string
```

Example:

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def __str__(self):
        return f"{self.name} scored {self.marks}"

s = Student("Vedant", 85)
print(s)
```

---

### 2️⃣ Arithmetic Operators

| Operator | Magic Method    |
| -------- | --------------- |
| `+`      | `__add__()`     |
| `-`      | `__sub__()`     |
| `*`      | `__mul__()`     |
| `/`      | `__truediv__()` |
| `%`      | `__mod__()`     |

Example:

```python
class Number:
    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        return self.value + other.value

a = Number(10)
b = Number(20)
print(a + b)  # 30
```

---

### 3️⃣ Comparison Operators

| Operator | Magic Method |
| -------- | ------------ |
| `==`     | `__eq__()`   |
| `!=`     | `__ne__()`   |
| `<`      | `__lt__()`   |
| `>`      | `__gt__()`   |
| `<=`     | `__le__()`   |
| `>=`     | `__ge__()`   |

Example:

```python
class Box:
    def __init__(self, volume):
        self.volume = volume

    def __gt__(self, other):
        return self.volume > other.volume
```

---

### 4️⃣ Length, Indexing & Iteration

| Feature      | Magic Method    |
| ------------ | --------------- |
| `len(obj)`   | `__len__()`     |
| `obj[i]`     | `__getitem__()` |
| `obj[i] = x` | `__setitem__()` |
| iteration    | `__iter__()`    |
| next item    | `__next__()`    |

Example:

```python
class MyList:
    def __init__(self, data):
        self.data = data

    def __len__(self):
        return len(self.data)

ml = MyList([1, 2, 3])
print(len(ml))
```

---

### 5️⃣ Callable Objects

```python
__call__()
```

Makes an object behave like a function.

```python
class Greet:
    def __call__(self, name):
        return f"Hello {name}"

g = Greet()
print(g("Vedant"))
```

---

### 6️⃣ Attribute Access Control

| Method               | Purpose                       |
| -------------------- | ----------------------------- |
| `__getattr__()`      | called if attribute not found |
| `__getattribute__()` | called for every attribute    |
| `__setattr__()`      | when setting attribute        |
| `__delattr__()`      | when deleting attribute       |

---

### 7️⃣ Context Managers (`with` statement)

```python
__enter__()
__exit__()
```

Example:

```python
class FileManager:
    def __enter__(self):
        print("File opened")
        return self

    def __exit__(self, exc_type, exc_val, exc_tb):
        print("File closed")
```

````python
# basic methods
class Person:
  def __init__(self,name,age):
    self.name = name
    self.age = age
    
person = Person("John",36)
print(person.name)
print(person.age)
````

````python
# overriding magic methods
class Person:
  def __init__(self,name,age):
    self.name = name
    self.age = age
    
  def __str__(self):
    return f"{self.name}({self.age})"
  
  def __repr__(self):
    return f"Person({self.name},{self.age})"
  
person = Person("John",36)
print(person)
print(repr(person))
````
