# Classes and Objects in Python

Object-oriented programming (OOP) is a programming paradigm that uses "objects" to design applications and programs. In Python, you can create classes and objects to implement OOP concepts.

## Classes
A class is a blueprint for creating objects. It defines a set of attributes and methods that the objects created from the class will have.

```python
class Dog:
    def __init__(self, name, age):
        self.name = name  # Attribute
        self.age = age    # Attribute

    def bark(self):  # Method
        return f"{self.name} says woof!"
```

In the example above, we defined a class called `Dog` with an `__init__` method (constructor) that initializes the attributes `name` and `age`. We also defined a method called `bark` that returns a string when called.

## Objects
An object is an instance of a class. You can create multiple objects from the same class, each with its own unique attributes.

```python
# Creating objects from the Dog class
dog1 = Dog("Buddy", 3)
dog2 = Dog("Max", 5)
# Accessing attributes and methods
print(dog1.name)  # Output: Buddy
print(dog1.bark())  # Output: Buddy says woof!
print(dog2.name)  # Output: Max

````python
# Instance variable and method 

class dog:
  # constructor __init_ is used to initialize the instance variable
  def __init__(self,name,age):
    # self is used to refer to the current instance of the class
    self.name=name
    self.age=age
    
# creating an object
dog1=dog("tommy",5)
print(dog1.name)
print(dog1.age)
````

````python
dog2=dog("Lucy",3)
print(dog2.name)
print(dog2.age)
````

````python
# Instance methods
class dog:
  def __init__(self,name,age):
    self.name=name
    self.age=age
    
  def bark(self):
    print(f"{self.name} says Woof! Woof!")

dog1=dog("tommy",5)
print(dog1.name)
print(dog1.age)
dog1.bark()
````

````python
# Modeling a bank account using OOPS

class BankAccount:
  def __init__ (self,owner,balance=0):
    self.owner=owner
    self.balance=balance
  def deposit(self,amount):
    self.balance+=amount
    print(f"Deposit of {amount} successful. New balance: {self.balance}")
    
  def withdraw(self,amount):
    if amount>self.balance:
      print("Insufficient funds")
    else:
      self.balance-=amount
      print(f"Withdrawal of {amount} successful. New balance: {self.balance}")
      
account1=BankAccount("Alice",1000)
print(account1.owner)
account1.deposit(500)
account1.withdraw(200)
````
